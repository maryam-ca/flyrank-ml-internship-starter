# 1. Research Question

## Research Question

Which content pages should be prioritized for refresh based on search performance, engagement metrics, content freshness, and historical trends?

## Decision Supported

This project provides a machine learning-based decision support system that identifies and ranks content pages according to their refresh priority. The objective is to help content teams focus their optimization efforts on pages with the highest potential impact.

This project uses the **FlyRank Internship Starter Dataset**, containing anonymized search performance and content quality signals.

### Dataset Summary

- Total Records: **30,000**

### Data Preparation

The following preprocessing steps were performed:

- Removed the **provider_used** column due to excessive missing values.
- Filled numerical missing values using the median.
- Filled categorical missing values using the mode.
- Verified that no duplicate records existed.
- Created additional features for refresh opportunity scoring.

## Refresh Opportunity Score

The refresh score was calculated using the following weighted framework:

- Content Age (30%)
- Trend Decline (25%)
- Click-Through Rate (20%)
- Engagement Rate (15%)
- Search Volume (10%)

The resulting score was scaled between 0 and 100 and categorized into:

- High Priority
- Medium Priority
- Low Priority

## 5. Limitations

*What this work cannot claim.*

# 5. Limitations

This project should be interpreted as a decision-support system rather than a causal ranking model.

The refresh priority labels were generated using a rule-based scoring framework built from observable search and content signals. Consequently, the machine learning models learn to approximate this framework rather than independently observed editorial decisions.

The findings identify patterns associated with refresh opportunities but do not claim to reveal or predict Google's ranking algorithm.

The analysis is based solely on the provided anonymized dataset and should be validated before applying recommendations in production environments.

# 6. Ranked Recommendations

Based on the Refresh Opportunity Scoring model, content pages were ranked into three priority levels to support editorial decision-making.

## High Priority
Pages in this category should be refreshed immediately because they typically exhibit one or more of the following characteristics:

- High content age
- Declining search trend
- Low click-through rate (CTR)
- Low engagement rate
- High search demand

**Recommended Actions**

- Update outdated content
- Rewrite titles and meta descriptions
- Improve content quality and readability
- Add new sections and FAQs
- Refresh internal links
- Improve keyword coverage

---

## Medium Priority

These pages are performing reasonably well but would benefit from scheduled optimization.

**Recommended Actions**

- Review metadata
- Improve content depth
- Update statistics and references
- Optimize headings and structure
- Monitor performance after updates

---

## Low Priority

These pages currently demonstrate healthy performance and require only routine monitoring.

**Recommended Actions**

- No immediate refresh required
- Monitor monthly
- Continue performance tracking
