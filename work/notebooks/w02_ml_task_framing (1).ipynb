## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*



My lane is **Ranking**.

The goal is not simply to classify search results as relevant or irrelevant. Instead, the system should rank multiple documents so that the most relevant content appears at the top of the search results.

For every search query, the machine learning model predicts how relevant each document is and assigns a score. The documents are then ordered from highest relevance to lowest relevance.

This is a ranking problem because users expect the best answer to appear first rather than just knowing whether a document is relevant.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*



The target is the **relevance score** between a user's search query and a document.

Since true relevance labels are not always available, we can use proxy signals such as:

- User clicks
- Time spent reading the page
- Search result position
- User engagement
- Manual relevance labels

The model learns to predict how useful a document is for a given search query.

## 3. Success metric

*One metric you can defend. What number means 'good'?*



The primary success metric is **NDCG (Normalized Discounted Cumulative Gain)**.

NDCG measures whether the most relevant documents appear near the top of the ranked list.

Other useful metrics include:

- Precision@10
- Recall@10
- Mean Reciprocal Rank (MRR)

A higher NDCG score means users receive better search results.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*



One row represents one **query-document pair**.

Each row contains:

- Search query
- Document or webpage
- Features describing the query and document
- Relevance label or score

Example:

| Query | Document | Clicks | BM25 Score | Relevance |
|-------|----------|---------|------------|-----------|
| machine learning | ML Tutorial | 21 | 7.6 | 3 |

The model predicts the relevance score for each query-document pair.

```python
import pandas as pd

data = {
    "query": [
        "machine learning",
        "python tutorial",
        "deep learning",
        "neural networks",
        "data science"
    ],
    "document": [
        "ML Guide",
        "Python Basics",
        "Deep Learning Book",
        "Neural Network Notes",
        "Data Science Handbook"
    ],
    "clicks": [35, 48, 25, 18, 42],
    "bm25_score": [8.5, 9.1, 7.8, 6.9, 8.8],
    "relevance": [3,3,2,2,3]
}

df = pd.DataFrame(data)

print(df)

df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*



A fixed rule cannot capture the complexity of human search behavior.

Different users may search using different words while expecting the same answer.

Machine learning can learn hidden relationships between queries and documents using many features such as:

- Keyword similarity
- Semantic similarity
- User clicks
- User behavior
- Historical relevance
- Document quality

As more search data becomes available, the ranking model continuously improves without manually writing new rules.

Therefore, ML provides more accurate, personalized, and scalable search ranking than a fixed rule-based system.
