## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

Rule:

Prioritize content that has not been updated for a long time, has low CTR, and receives a reasonable number of impressions. These pages are good candidates for content refresh because improving them may increase clicks and engagement.

Reason Code:

STALE_LOW_CTR

```python
# Signal Check 1: CTR Buckets

df["ctr_bucket"] = pd.cut(
    df["ctr"],
    bins=[-0.01, 0.1, 1, 5, 100],
    labels=["Very Low", "Low", "Medium", "High"]
)

ctr_table = (
    df.groupby("ctr_bucket")
      .agg(
          n=("ctr", "count"),
          avg_position=("avg_position", "mean"),
          avg_impressions=("impressions_90d", "mean")
      )
)

print("Signal 1: CTR Bucket Table")
display(ctr_table)
```

Signal 1 Verdict: CONFIRMED

Reason:

Very low CTR pages generally have weaker performance and are suitable candidates for content improvement.

```python
# Signal Check 2: Days Since Last Update (Freshness)

df["update_bucket"] = pd.cut(
    df["days_since_last_update"],
    bins=[0, 30, 90, 180, 365, 1000],
    labels=[
        "0-30 Days",
        "31-90 Days",
        "91-180 Days",
        "181-365 Days",
        "365+ Days"
    ]
)

freshness_table = (
    df.groupby("update_bucket")
      .agg(
          n=("days_since_last_update", "count"),
          avg_ctr=("ctr", "mean"),
          avg_impressions=("impressions_90d", "mean")
      )
)

print("Signal 2: Freshness Bucket Table")
display(freshness_table)
```

Signal 2 Verdict: CONFIRMED

Reason:

Older content generally has lower CTR and is a good candidate for content refresh. Pages that have not been updated for a long time should be reviewed first.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# Copy dataset
baseline = df.copy()

# Initial Score
baseline["score"] = 0
```

```python
baseline.loc[baseline["ctr"] < 0.1, "score"] += 40
baseline.loc[(baseline["ctr"] >= 0.1) & (baseline["ctr"] < 1), "score"] += 25
```

```python
baseline.loc[baseline["days_since_last_update"] > 180, "score"] += 30
baseline.loc[
    (baseline["days_since_last_update"] > 90)
    & (baseline["days_since_last_update"] <= 180),
    "score"
] += 20
```

```python
baseline.loc[baseline["impressions_90d"] > 5000, "score"] += 30
baseline.loc[
    (baseline["impressions_90d"] > 1000)
    & (baseline["impressions_90d"] <= 5000),
    "score"
] += 20
```

```python
baseline["reason_code"] = "STALE_LOW_CTR"
```

```python
baseline["action"] = "Refresh Content"
```

```python
baseline = baseline.sort_values(
    by="score",
    ascending=False
)

baseline.head(10)
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("work/outputs", exist_ok=True)

# Save ranked queue
baseline.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print("✅ CSV Saved Successfully!")
```
