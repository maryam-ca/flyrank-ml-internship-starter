## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*



The dataset provides useful search intelligence but also has several limitations.

1. The dataset is anonymized.

Client names and page URLs have been removed.

2. Missing values exist.

Several columns such as provider_used, model_used, word_count and char_count contain missing values.

3. No explicit target variable exists.

A prediction label must be created manually.

4. Historical data only.

The dataset summarizes past performance and cannot directly predict future behaviour.

5. External factors are unavailable.

Google algorithm updates, competitor activity and seasonal events are not included.

6. Trend values are observational.

A negative trend does not always mean poor content quality.

7. Correlation does not imply causation.

Higher impressions or CTR alone cannot explain why performance changed.

Therefore, results should support human decision-making rather than replace it.
